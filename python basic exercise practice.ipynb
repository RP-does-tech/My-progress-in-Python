```python
#Count the vowels Exercise 16
word = input("Enter a word: ")
count = 0
for letter in word:
    if letter in "aeiou" :
        count = count + 1
print("Vowels:", count)
```

```python
numbers = [3, 8, 12, 5, 7, 10, 21, 4]

even_count = 0
odd_count = 0

for number in numbers:
    if number % 2 == 0:
        even_count = even_count + 1
    else:
        odd_count = odd_count + 1

print("Even:", even_count)
print("Odd:", odd_count)
```
