```python
def calculate_average(score1, score2): 
    average = (score1 + score2) / 2 
    return average 
result = calculate_average(70, 80,)
if result > 70 :
    print(result, "Good")
else :
    print(result, "Bad")
```

```python
def contains_query(document, query): 
    return query.lower() in document.lower() 

print(contains_query("Introduction to Information Retrieval", 
"retrieval"))

print(contains_query("Data Visualization", "retrieval"))
```

```python
def classify_score(score):
    if score >= 80 :
        return "High" 
    elif score >= 50 :
        return "Moderate"
    else :
        return "Low" 
print(classify_score(92))
print(classify_score(68))
print(classify_score(40))
```
