```python
# Find the largest number Exercise 32

numbers = [12, 45, 7, 23, 56, 18, 9]

largest = numbers[0]

for number in numbers:
    if number > largest:
        largest = number

print("Largest number:", largest)
```

```python
#Find the Smallest number Exercise 33

numbers = [15, 8, 23, 4, 19, 7, 12]

smallest = numbers[0]
for number in numbers:
    if number < smallest:
        smallest = number

print("Smallest number:", smallest)
```

```python
#Count the numbers greater than 10 Exercise 34

numbers = [5, 12, 18, 7, 25, 3, 15, 9]

count = 0

for number in numbers:
    if number > 10 :
        count +=1
print("Number greater than 10:", count)
```

```python
#Count the even numbers Exercise 35

numbers = [3, 8, 12, 5, 7, 10, 21, 4]

count = 0

for number in numbers:
    if number % 2 == 0:
        count +=1

print("Even numbers:", count)
```

```python
numbers = [3, 8, 12, 5, 7, 10, 21, 4]

count = 0 

for number in numbers:
    if number % 2 == 1:
        count +=1

print("Odd numbers:", count)

```
