```python
documents = [ 
"Artificial intelligence is transforming education", 
"Information retrieval helps users find relevant documents", 
"Python is widely used for data analysis", 
"Data visualization communicates information visually", 
"Search engines use information retrieval techniques", 
"Machine learning can identify patterns in data" 
]

query = "information"
query1 = "data"
query2 = "Python"

#Sedarching for "information"
match_count = 0 
for document in documents:
    if query in document.lower():
        print(document)
        match_count += 1

if match_count == 0:
    print("No document found")

print("Number of counts:", match_count)

#Searching for "data"
match_count = 0
for document in documents:
    if query1 in document.lower():
        print(document)
        match_count += 1

if match_count == 0:
    print("No document found")
 
print("Number of counts:", match_count)

#Serarching for "Python"
match_count = 0
for document in documents:
    if query2 in document.lower():
        print(document)
        match_count += 1

if match_count == 0:
    print("No document found")

print("Number of counts:", match_count)
```
