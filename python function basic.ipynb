```python
def greet():
    return "Hello, welcome to Python!"

print(greet())
```

```python
def greet_user(name):
    return "Hello " + name + "!"

print(greet_user("Alex"))
```

```python
def add_numbers (number1, number2) :
    return number1 + number2

print(add_numbers(5, 3))
```

```python
def check_even(number):
    if number % 2 == 0:
        return "Even"
    else:
        return "Odd"

print(check_even(8))
print(check_even(7))
```

```python
def square(number) :
    return number * number 

print(square(5))
```

```python
def find_larger(a, b):
    if a > b:
        return a
    else:
        return b

print(find_larger(10, 5))
print(find_larger(3, 8))
```

```python
numbers = [3, 8, 12, 5, 7, 10]

def count_even(numbers):
    count = 0

    for number in numbers:
        if number % 2 == 0:
            count += 1

    return count

print(count_even(numbers))
```
