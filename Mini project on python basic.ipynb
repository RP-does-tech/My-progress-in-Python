```python
numbers = [3, 8, 12, 5, 7, 10, 21, 4]

even_count = 0
odd_count = 0
greater_than_10 = 0

for number in numbers:

    # Check if number is even
    if number % 2 == 0:
        even_count +=1


    # Check if number is odd
    if number % 2 == 1:
        odd_count +=1
        


    # Check if number is greater than 10
    if number >= 10:
        greater_than_10 +=1


print("Total numbers:", len(numbers))
print("Even numbers:", even_count)
print("Odd numbers:", odd_count)
print("Greater than 10:", greater_than_10)
```

```python
numbers = [12, 5, 27, 8, 19, 3, 15]

largest = numbers[0]

for number in numbers:
    if number > largest :
        largest = number

print("Largest number:", largest)
```
