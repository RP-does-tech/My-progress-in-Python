```python
#Find the largest number Exercise 18
numbers = [12,5,27,8,19,3]

largest = 0

for number in numbers :
    if number >= largest :
        largest = number

print("Largest:", largest)
```

```python
#Find the smallest number Exercise 19

numbers = [15,7,22,4,18,9]

smallest = numbers[0]

for number in numbers :
    if number <= smallest :
        smallest = number
print("Smallest:", smallest)
```

```python
# Find the average - Exercise 24

numbers = [10, 20, 30, 40, 50]

total = 0

for number in numbers:
    if number > 0 :
        total += number

average = total / 5

print("Average:", average)
```

```python
# Count even numbers - Exercise 25

numbers = [3, 8, 12, 5, 7, 10, 21, 4]

count = 0

for number in numbers:
    if number % 2 == 0 :
        count +=1

print("Even numbers:", count)
```

```python
# Sum of even numbers - Exercise 27

numbers = [3, 8, 12, 5, 7, 10, 21, 4]

total = 0

for number in numbers:
    if number % 2 == 0 :
        total += number

print("Sum of even numbers:", total)
```

```python
# Sum of odd numbers - Exercise 28

numbers = [3, 8, 12, 5, 7, 10, 21, 4]

total = 0

for number in numbers:
    if number % 2 == 1 :
        total += number

print("Sum of odd numbers:", total)
```

```python
# Find the largest even number - Exercise 29

numbers = [3, 8, 12, 5, 7, 10, 21, 4]

largest = 0

for number in numbers:
    if number % 2 == 0 :
        if number > largest :
            largest = number

print("Largest even number:", largest)
```

```python
# Find the smallest odd number - Exercise 30

numbers = [12, 7, 15, 4, 9, 20, 3, 8]

smallest = numbers[0]

for number in numbers:
    if number % 2 == 1 :
        if number < smallest :
            smallest = number

print("Smallest odd number:", smallest)
```

```python
# Count numbers between 10 and 20 - Exercise 31

numbers = [5, 12, 18, 25, 10, 7, 15, 22, 20]

count = 0

for number in numbers:
    if number >= 10 and number <= 20 :
        count += 1

print("Numbers between 10 and 20:", count)
```
